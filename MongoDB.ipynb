````python


**1. What are the key differences between SQL and NoSQL databases?**

* **Structure:** SQL databases are relational and use structured tables with fixed schemas. NoSQL databases are non-relational and store data in flexible formats like JSON-like documents, key-value pairs, or graphs.
* **Scaling:** SQL databases typically scale vertically (adding more power to a single server). NoSQL databases are designed to scale horizontally (adding more servers across a cluster).
* **Transactions:** SQL databases natively support complex multi-row ACID transactions. While MongoDB now supports multi-document ACID transactions, NoSQL generally prioritizes performance and availability over strict transactional consistency across distributed nodes.

**2. What makes MongoDB a good choice for modern applications?**


MongoDB uses a flexible, JSON-like document model (BSON) that maps naturally to objects in modern programming languages. It allows developers to iterate rapidly without migrating rigid schemas. It also offers built-in high availability, horizontal scaling, and powerful indexing/aggregation capabilities.

**3. Explain the concept of collections in MongoDB.**


A collection in MongoDB is the equivalent of a table in a relational database. It is a grouping of MongoDB documents. Unlike SQL tables, collections do not enforce a strict schema, meaning documents within the same collection can have different fields and structures.

**4. How does MongoDB ensure high availability using replication?**


MongoDB uses "Replica Sets," which are groups of MongoDB servers that maintain the same data set. One node acts as the Primary (handling write operations), and the others are Secondaries (replicating the Primary's data). If the Primary fails, an automated election process promotes a Secondary to become the new Primary, ensuring continuous availability without manual intervention.

**5. What are the main benefits of MongoDB Atlas?**


MongoDB Atlas is the fully managed cloud database service for MongoDB. Benefits include automated backups, built-in security features (VPC peering, IP whitelisting, encryption), global clustering, auto-scaling, and the removal of administrative overhead (like patching and provisioning).

**6. What is the role of indexes in MongoDB, and how do they improve performance?**


Indexes are special data structures that store a small portion of the collection's data set in an easy-to-traverse form. Without indexes, MongoDB must perform a "collection scan" (reading every document) to find matches. Indexes allow the database engine to quickly locate and return the requested documents, drastically improving read performance.

**7. Describe the stages of the MongoDB aggregation pipeline.**


The aggregation pipeline is a framework for data processing where documents enter a multi-stage pipeline that transforms them into an aggregated result. Common stages include:

* `$match`: Filters documents.
* `$group`: Groups documents by a specified key and applies accumulator expressions (e.g., sum, avg).
* `$sort`: Orders the documents.
* `$project`: Reshapes documents (adds, removes, or renames fields).
* `$limit` / `$skip`: Controls the number of documents passed forward.

**8. What is sharding in MongoDB? How does it differ from replication?**


Sharding is the process of distributing data across multiple machines to support horizontal scaling (handling massive datasets and high throughput).

* **Sharding** partitions data across multiple servers so no single server holds all the data (increases capacity/speed).
* **Replication** copies the *same* data across multiple servers (increases availability/fault tolerance).

**9. What is PyMongo, and why is it used?**


PyMongo is the official, native Python driver for MongoDB. It is used to write Python applications that interact with MongoDB clusters, allowing developers to execute queries, insert data, and manage the database directly through Python code.

**10. What are the ACID properties in the context of MongoDB transactions?**


ACID stands for Atomicity, Consistency, Isolation, and Durability. Since version 4.0, MongoDB supports multi-document ACID transactions, meaning you can update multiple documents across multiple collections, and either all operations succeed completely (commit), or none of them apply at all (rollback), maintaining data integrity.

**11. What is the purpose of MongoDB's explain() function?**


The `explain()` function returns detailed information about how MongoDB executes a query. It provides the "execution plan," revealing whether an index was used, how many documents were examined, and how long the query took. It is a critical tool for performance tuning.

**12. How does MongoDB handle schema validation?**


MongoDB can enforce schema validation rules at the collection level using JSON Schema. When validation is configured, MongoDB checks inserted or updated documents against the defined rules (e.g., ensuring a `price` field is a positive number, or an `email` field exists) and rejects the operation if it fails.

**13. What is the difference between a primary and a secondary node in a replica set?**

* **Primary Node:** The only member in the replica set that receives write operations. It records all changes to its data sets in its oplog.
* **Secondary Node:** Replicates the primary's oplog and applies the operations to its own data set asynchronously. Secondaries can be configured to serve read operations to distribute workload.

**14. What security mechanisms does MongoDB provide for data protection?**


MongoDB provides robust security including Role-Based Access Control (RBAC) to enforce least privilege, Authentication (SCRAM, x.509, LDAP), Network Isolation (IP whitelisting, VPCs), TLS/SSL encryption for data in transit, and Field Level Encryption / Encryption at Rest for stored data.

**15. Explain the concept of embedded documents and when they should be used.**


Embedded documents capture relationships between data by storing related data in a single document structure (nesting objects or arrays). They should be used for "contains" or "one-to-many" relationships where the nested data is frequently accessed at the same time as the parent document, minimizing the need for complex joins.

**16. What is the purpose of MongoDB's $lookup stage in aggregation?**


The `$lookup` stage performs a left outer join to an unsharded collection in the same database. It allows you to filter in documents from the "joined" collection for processing, combining related data across multiple collections.

**17. What are some common use cases for MongoDB?**


Common use cases include product catalogs, Content Management Systems (CMS), Internet of Things (IoT) telemetry data storage, real-time analytics, user profiles/session management, and mobile application backends.

**18. What are the advantages of using MongoDB for horizontal scaling?**


MongoDBs native sharding architecture allows you to scale horizontally seamlessly. The advantages include cost savings (using multiple cheaper servers instead of one massive mainframe), lack of a single point of failure, and the ability to dynamically add capacity as data volume and query traffic grow.

**19. How do MongoDB transactions differ from SQL transactions?**


While both now support ACID transactions, SQL transactions are a fundamental, everyday requirement due to normalized, multi-table schemas. MongoDB transactions are an exception rather than the rule; because of the document model, most operations can be resolved with a single-document atomic update, making multi-document transactions necessary only in specific, complex edge cases.

**20. What are the main differences between capped collections and regular collections?**


Capped collections are fixed-size collections that support high-throughput operations based on insertion order. Once the collection fills its allocated space, it automatically overwrites the oldest documents (like a circular buffer). Regular collections can grow dynamically in size and do not automatically delete old documents.

**21. What is the purpose of the $match stage in MongoDB's aggregation pipeline?**


The `$match` stage filters the document stream, allowing only matching documents to pass unmodified into the next pipeline stage. It acts exactly like the standard MongoDB `find()` query and is usually placed early in the pipeline to reduce the amount of data processed.

**22. How can you secure access to a MongoDB database?**


Access is secured by enforcing authentication, creating specific users with Role-Based Access Control (RBAC), binding MongoDB to secure IP addresses, enabling TLS/SSL for network connections, and enabling encryption at rest.

**23. What is MongoDB's WiredTiger storage engine, and why is it important?**


WiredTiger is the default storage engine for MongoDB. It is important because it provides document-level concurrency control (allowing multiple clients to modify different documents in a collection simultaneously), highly efficient data compression (reducing storage costs), and optimal memory utilization.





import pandas as pd
from pymongo import MongoClient

# Establish connection to MongoDB (Assuming a local instance or replace with your Atlas URI)
# client = MongoClient("mongodb+srv://<username>:<password>@cluster.mongodb.net/")
client = MongoClient("mongodb://localhost:27017/")

# Create/Connect to the database and collection
db = client["SuperstoreDB"]
collection = db["Orders"]

# Practical 1: Write a Python script to load the Superstore dataset from a CSV file into MongoDB.[cite: 4]
# Make sure "Superstore.csv" is uploaded to your Colab environment
try:
    df = pd.read_csv("Superstore.csv")
    records = df.to_dict(orient='records')

    # Clear the collection first if running multiple times, then insert
    collection.delete_many({})
    collection.insert_many(records)
    print("Dataset successfully loaded into MongoDB.")
except FileNotFoundError:
    print("Please upload 'Superstore.csv' to the Colab files pane to run this block.")



```python
# Practical 2: Retrieve and print all documents from the Orders collection.[cite: 4]
# (Using limit(3) to prevent crashing the Colab output with massive data)
all_orders = collection.find().limit(3)
print("--- All Documents (Preview of first 3) ---")
for order in all_orders:
    print(order)

# Practical 3: Count and display the total number of documents in the Orders collection.[cite: 4]
total_docs = collection.count_documents({})
print(f"\nTotal documents in Orders collection: {total_docs}")

# Practical 4: Write a query to fetch all orders from the "West" region.[cite: 4]
west_orders = collection.find({"Region": "West"})
print(f"\nTotal orders from West region: {collection.count_documents({'Region': 'West'})}")

# Practical 5: Write a query to find orders where Sales is greater than 500.[cite: 4]
high_sales_orders = collection.find({"Sales": {"$gt": 500}})
print(f"Total orders with Sales > 500: {collection.count_documents({'Sales': {'$gt': 500}})}")



```python
# Practical 6: Fetch the top 3 orders with the highest Profit.[cite: 4]
top_3_profit = collection.find().sort("Profit", -1).limit(3)
print("--- Top 3 Orders by Highest Profit ---")
for order in top_3_profit:
    print(f"Order ID: {order.get('Order ID', 'N/A')} | Profit: {order.get('Profit', 0)}")

# Practical 7: Update all orders with Ship Mode as "First Class" to "Premium Class."[cite: 4]
update_result = collection.update_many(
    {"Ship Mode": "First Class"},
    {"$set": {"Ship Mode": "Premium Class"}}
)
print(f"\nDocuments updated to Premium Class: {update_result.modified_count}")

# Practical 8: Delete all orders where Sales is less than 50.[cite: 4]
delete_result = collection.delete_many({"Sales": {"$lt": 50}})
print(f"Documents deleted (Sales < 50): {delete_result.deleted_count}")



```python
# Practical 9: Use aggregation to group orders by Region and calculate total sales per region.[cite: 4]
pipeline = [
    {"$group": {
        "_id": "$Region",
        "TotalSales": {"$sum": "$Sales"}
    }},
    {"$sort": {"TotalSales": -1}}
]
sales_by_region = collection.aggregate(pipeline)

print("--- Total Sales per Region ---")
for result in sales_by_region:
    print(f"Region: {result['_id']} | Total Sales: ${result['TotalSales']:.2f}")

# Practical 10: Fetch all distinct values for Ship Mode from the collection.[cite: 4]
distinct_ship_modes = collection.distinct("Ship Mode")
print(f"\nDistinct Ship Modes: {distinct_ship_modes}")

# Practical 11: Count the number of orders for each category.[cite: 4]
category_pipeline = [
    {"$group": {
        "_id": "$Category",
        "OrderCount": {"$sum": 1}
    }},
    {"$sort": {"OrderCount": -1}}
]
category_counts = collection.aggregate(category_pipeline)

print("\n--- Order Count per Category ---")
for category in category_counts:
    print(f"Category: {category['_id']} | Orders: {category['OrderCount']}")

```
````
